```python
data1.drop("Unnamed: 0",axis=1,inplace = True)
data1
```

```python
others_data1= pd.DataFrame({
            "others_data":data1.iloc[:,-1].apply(lambda x:x.split(","))})
others_data1
```

```python
type(others_data1)
```

```python
indus_s = set()
def old(x):
    len_x = len(x)
    for i in range(len_x):
        if "years" in x[i]:
            return x[i]
    else:
        return np.nan
```

```python
others_data1 = others_data1.assign(
old = others_data1.loc[:,"others_data"].apply(old)
                                  )
others_data1
```

```python
index = others_data1[~(others_data1["old"].isna())]
index
```

```python
def rem_years(x):
    len_x = len(x)
    for i in range(len_x):
        if "years" in x[i]:
            return x[:i]+x[i+1:]
```

```python
others_data1["others_data"] = pd.concat((index["others_data"].apply(rem_years),others_data1[others_data1["old"].isna()]["others_data"]),axis = 0)
others_data1["others_data"]
```

```python
others_data1
```

```python
def employees(x):
    len_x = len(x)
    for i in range(len_x):
        if "employees" in x[i].lower():
            return x[i]
    else:
        return np.nan
```

```python
others_data1 = others_data1.assign(
total_employees = others_data1.loc[:,"others_data"].apply(employees)
                                )
others_data1
```

```python
def remove_employ(x):
    len_x = len(x)
    for i in range(len_x):
        if "employees" in x[i].lower():
            return x[:i]+x[i+1:]
    else:
        return x
```

```python
others_data1["others_data"] = others_data1["others_data"].apply(remove_employ)
others_data1["others_data"]
```

```python
others_data1
```

```python
city = set()
def cities(x):
    len_x = len(x)
    for i in range(len_x):
        if "+" in x[i] or x[i].strip().lower() in city:
            city.add(x[i].split("+")[0].strip().lower())
            return x[i]
    else:
        return np.nan
```

```python
others_data1 = others_data1.assign(
cities = others_data1.loc[:,"others_data"].apply(cities)
                                )
others_data1
```

```python
def remove_cities(x):
    len_x = len(x)
    for i in range(len_x):
        if x[i].lower().strip() in city or "+" in x[i] :
            return x[:i]+x[i+1:]
    return x
```

```python
others_data1["others_data"] = others_data1["others_data"].apply(remove_cities)
others_data1["others_data"]
```

```python
others_data1
```

```python
ty = ["public","forbes global 2000", "conglomerate", "startup","fortune global 500","fortune india 500","fortune us 500", "indian unicorn", "government"]
def typ(x):
    len_x = len(x)
    for i in range(len_x):
        if x[i].strip().lower() in ty:
            return x[i]
    else:
        return np.nan
```

```python
others_data1 = others_data1.assign(
type = others_data1.loc[:,"others_data"].apply(typ)
                                )
others_data1
```

```python
ty = ["public","forbes global 2000", "conglomerate", "startup","fortune global 500","fortune india 500","fortune us 500", "indian unicorn", "government"]
def remove_typ(x):
    len_x = len(x)
    for i in range(len_x):
        if x[i].strip().lower() in ty:
            return x[:i]+x[i+1:]
    else:
        return []
```

```python
others_data1["others_data"] = others_data1["others_data"].apply(remove_typ)
others_data1["others_data"]
```

```python
others_data1
```

```python
def indus(x):
    if x == []:
        return np.nan
    return x[0]
```

```python
others_data1 = others_data1.assign(
industry = others_data1.loc[:,"others_data"].apply(indus)
                                )
others_data1
```

```python
others_data = others_data1.drop(labels = "others_data",axis =1)
others_data
```

```python
data1 = pd.concat([data1.iloc[:,:2],others_data1.iloc[:,1:]],axis=1)
data1
```
